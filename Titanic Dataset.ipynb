```python
import numpy as np
a=df.select_dtypes(include=np.number)
```

```python
for i in a:
    q1=a[i].quantile(0.25)
    q2=a[i].quantile(0.50)
    q3=a[i].quantile(0.75)
    IQR=q3-q1
    Uw=q3+1.5*IQR
    Lw=q1-1.5*IQR
    outliers=((a[i]>Uw)|(a[i]<Lw))
    print(f'{i}:-',outliers.sum())
    a.loc[outliers,i]=q2
    print(((a[i]>Uw)|(a[i]<Lw)).sum())
    
    
```
